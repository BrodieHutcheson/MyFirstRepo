## Part 2: Mapping and analysing traffic sensor distribution across Glasgow
For this challenge I used the Glasglow Open Data API ( Transit) https://developer.glasgow.gov.uk/api-details#api=traffic&operation=traffic-sensor-locations to gather all the sensor locations in the city. The sensor data was converted into a geodataframe and then mapped. Data zone boundaries where then spatially joined with the sensor geodataframe to visualise which sensors fell within the data zone polygons. Groupby() was used to count the number of sensors per data zone using the data zone code

### Data sources

- [Glasgow traffic movement sites API](https://api.glasgow.gov.uk/traffic/v1/movement/sites)
- [Scottish Government 2022 Data Zones metadata](https://spatialdata.gov.scot/geonetwork/srv/eng/catalog.search#/metadata/f6656adf-b720-4612-ad5c-1d13eae94c8b)
- [Scottish local authority boundaries](https://www.data.gov.uk/dataset/8e3a4564-8081-42ec-8772-03ade11d4acf/local-authority-boundaries-scotland1)

```python
import requests  # imports the library used to request and receive data from APIs
import pandas as pd
import geopandas as gpd
```

```python
sensors = "https://api.glasgow.gov.uk/traffic/v1/movement/sites"  # store API URL under sensors variable

response = requests.get(sensors)  # send a request to the API and store the response
response
data = response.json()  # converts API response from JSON to python format and stores as 'data' variable
data
```

```python

site_ids = []  # create empty lists to store site IDs and coordinates
from_lats = []
from_longs = []
from_descriptions = []


for entry in data:  # use for loop to go through API sensor records and extract the site IDs, coordinates and descriptions
    site_id = entry['siteId']
    from_description = entry['from']['description']
    from_lat = entry['from']['lat']
    from_long = entry['from']['long']
    
    
    site_ids.append(site_id)  # add the extracted information to the following lists 
    from_descriptions.append(from_description)
    from_lats.append(from_lat)
    from_longs.append(from_long)
    

df = pd.DataFrame({  # add the lists to a pandas dataframe
    'siteId': site_ids,
    'from_description': from_descriptions,
    'from_lat': from_lats,
    'from_long': from_longs,
})
```

```python
df.head()  # display the 1st 5 rows of the dataframe
```

```python
gdf_sensors = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df['from_long'], df['from_lat']), crs='EPSG:4326')  # convert into a geodataframe
gdf_sensors.head()  # display the top of the geodataframe                                                                                                 # create geometries from the coordinates
                                                                                                                   # set the crs
```

```python
gdf_sensors.explore()  # map the geodataframe
```

```python
data_zones = gpd.read_file('Week 2 data/SG_DataZone_Bdry_2022.shp')  # read in the datazone boundary shapefile and store it within the 'data_zones' variable
```

```python
data_zones.info()  # summarise the dataframe
```

```python
print(data_zones.columns)  # display the columns of the dataframe
```

```python
print(data_zones.crs)  # print the crs of the dataframe
```

```python
data_zones.head()  # display the top of the dataframe
```

```python
local_authorities = gpd.read_file('Week 2 data/pub_las.shp')  # read in the local authority boundaries shapefile and store it within the 'local_authorities' variable
    

print(local_authorities.columns)  # display the columns within the dataframe
print(local_authorities.crs)  # show the crs of the dataframe
local_authorities.head()  # display the top of the dataframe
```

```python
glasgow_boundary = local_authorities.loc[  
    local_authorities["local_auth"] == "Glasgow City"]  # use .loc to extract all rows within the local authority column called 'Glasgow City'

glasgow_boundary.head()  # display the rows that are left after extracting
```

```python
glasgow_boundary = glasgow_boundary.to_crs(data_zones.crs)

# Select the data zones that intersect Glasgow City
glasgow_zones = data_zones.loc[
    data_zones.intersects(
        glasgow_boundary.unary_union)
    
].copy()

# Check the new GeoDataFrame
print(glasgow_zones.shape)
glasgow_zones.head()
```

```python
print(gdf_sensors.crs)
print(glasgow_zones.crs)  # show the crs of both geodataframes
```

```python
gdf_sensors = gdf_sensors.to_crs(27700)  # set the sensors gdf to the same as glasgow_zones
```

```python
print(gdf_sensors.crs)
print(glasgow_zones.crs)  # show that they are now the same
```

```python
sensors_in_zones = gpd.sjoin(  # using sjoin to spatially join the traffic sensor points to the Glasgow data zones polygon 
    gdf_sensors,  # sensor geodataframe
    glasgow_zones,  # glasgow data zones geodataframe
    how="inner",
    predicate="within"  # defines the spatial relaionship required for a match (sensor within polygon)
)
```

```python
sensors_in_zones.head()  # show the top of the new combined geodataframe
```

```python
sensor_counts = sensors_in_zones.groupby("dzcode").size()  # use groupby to group the joined sensor records by data zone to count the number of sensors per working zone
sensor_counts.head()  # show number of sensors per data zone code
```

```python
glasgow_zones = glasgow_zones.merge(  # merge sensor counts to the glasgow data zones gdf
    sensor_counts.rename("sensor_count"),  # rename to sensor count
    on="dzcode",  # match rows using the data zone column of the gdf
    how="left"
)  
```

```python
glasgow_zones.explore(  # map the number of sensors in each Glasgow data zone
    column="sensor_count",  # use the sensor count column
    cmap="YlOrRd",
    legend=True  # add legend
)
```

## Challenge 1: Define a function for calculating descriptive statistics
For this challenge I defined a function to calculate descriptive statistics for a dataframe. The dataframe included non-numerical columns so the function had to include only numerical attributes. The function returned a dataframe with the mean and counts for each numerical column

### Data source

The Airbnb listings data was obtained from [Airbnb Amsterdam listings](https://insideairbnb.com/get-the-data/).

```python
import pandas as pd

listings =pd.read_csv("Week 3 data/listings.csv")


#Show column headings
listings.head()
```

```python
subset_listings = listings[[
    "id",
    "neighbourhood",
    "latitude",
    "longitude",
    "room_type",
    "minimum_nights",
    "price",
    "number_of_reviews"
]].copy()

# Convert price to a numerical data type
subset_listings.loc[subset_listings.index, "price"] = pd.to_numeric(
    subset_listings["price"].replace(r"[\$,]", "", regex=True),
    errors="coerce"
)

# Remove records with missing values
subset_listings = subset_listings.dropna()

# Display the cleaned data
subset_listings.head()
  
```

```python
def data_description(df):  # defines a function for calculating descriptive statistics
    numeric_df = df.select_dtypes(include = 'number')  # selects only numerical columns
    stats = {'Mean': numeric_df.mean(), 'Counts': numeric_df.count()}
    return pd.DataFrame(stats)  # return the results as a dataframe

result = data_description(subset_listings)  # apply the function to the listings dataframe
print(result)  # print the descriptive statistics
```

The analysis uses the New York City Motor Vehicle Collisions — Crashes dataset. The API request is filtered to January 2024.

```python
import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
import pydeck as pdk
```

```python
url = "https://data.cityofnewyork.us/resource/h9gi-nx95.json"

params = {
    "$where": "crash_date >= '2024-01-01T00:00:00' AND crash_date < '2024-02-01T00:00:00'",
}

response = requests.get(url, params = params)
data = response.json()
crashes = pd.DataFrame(data)

crashes.head()
```

```python
columns = [
    "latitude",
    "longitude",
    "number_of_persons_injured",
    "number_of_persons_killed",
    "number_of_cyclist_injured",
    "number_of_cyclist_killed"
]

for col in columns:
    crashes[col] = pd.to_numeric(crashes[col],)

crashes[columns].head()
```

```python
crashes_map = crashes.dropna(
    subset=["latitude", "longitude"]
).copy()
```

```python
categories = [
    "No fatalities",
    "Person fatalities",
    "Cyclist fatalities"
]

colours = [
    [130, 130, 130, 70],
    [220, 0, 0, 200],
    [150, 0, 200, 220]
]

colour_dict = dict(zip(categories, colours))

colour_dict
```

```python
crashes_map["person_category"] = "No fatalities"

crashes_map.loc[
    crashes_map["number_of_persons_killed"] > 0,
    "person_category"
] = "Person fatalities"

crashes_map["person_colour"] = [
    colour_dict[category]
    for category in crashes_map["person_category"]
]
```

```python
crashes_map["cyclist_category"] = "No fatalities"

crashes_map.loc[
    crashes_map["number_of_cyclist_killed"] > 0,
    "cyclist_category"
] = "Cyclist fatalities"

crashes_map["cyclist_colour"] = [
    colour_dict[category]
    for category in crashes_map["cyclist_category"]
]
```

```python
fatal_person_crashes = crashes_map.loc[
    crashes_map["number_of_persons_killed"] > 0
].copy()

fatal_cyclist_crashes = crashes_map.loc[
    crashes_map["number_of_cyclist_killed"] > 0
].copy()
```

```python
cyclist_layer = pdk.Layer(
    "ScatterplotLayer",
    data=crashes_map,
    get_position="[longitude, latitude]",
    get_fill_color=colour_dict["Cyclist fatalities"],
    get_radius=35,
    radius_min_pixels=2,
    pickable=True
)

person_layer = pdk.Layer(
    "ScatterplotLayer",
    data=fatal_person_crashes,
    get_position="[longitude, latitude]",
    get_fill_color=colour_dict["Person fatalities"],
    get_radius=90,
    radius_min_pixels=3,
    pickable=True
)
```

```python
view_state = pdk.ViewState(
    longitude=crashes_map["longitude"].mean(),
    latitude=crashes_map["latitude"].mean(),
    zoom=10,
    pitch=0
)
```

```python
combined_map = pdk.Deck(
    layers=[
        person_layer,
        cyclist_layer
    ],
    initial_view_state=view_state
)

combined_map
```
